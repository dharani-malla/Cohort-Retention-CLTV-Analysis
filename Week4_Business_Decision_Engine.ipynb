##Import Required Libraries

```python
import pandas as pd
```

##Load Business KPI Dashboard

```python
performance = pd.read_excel("Business_KPI_Dashboard.xlsx")
print("Business KPI Dashboard Loaded Successfully")
display(performance)
```

##Rule-Based Recommendation Engine

```python
def generate_recommendation(row):

    # Highest Revenue but not Highest CLTV
    if row["Revenue_Rank"] == 1 and row["CLTV_Rank"] != 1:

        recommendation = "Improve customer retention through loyalty programs."

        priority = "High"

        impact = "Increase repeat purchases and long-term customer value."

    # Highest CLTV but not Highest Revenue
    elif row["CLTV_Rank"] == 1 and row["Revenue_Rank"] != 1:

        recommendation = "Increase customer acquisition through targeted marketing."

        priority = "High"

        impact = "Acquire more high-value customers."

    # Lowest Revenue Region
    elif row["Revenue_Rank"] == performance["Revenue_Rank"].max():

        recommendation = "Launch promotional campaigns and regional offers."

        priority = "Medium"

        impact = "Improve regional revenue."

    # Lowest Purchase Frequency
    elif row["Purchase_Frequency"] == performance["Purchase_Frequency"].min():

        recommendation = "Introduce personalized engagement campaigns."

        priority = "Medium"

        impact = "Increase purchase frequency."

    else:

        recommendation = "Continue monitoring customer behavior and optimize business strategies."

        priority = "Low"

        impact = "Maintain consistent business performance."

    return pd.Series([recommendation, priority, impact])
```

##Apply Recommendation Engine

```python
performance[
    ["Recommendation",
     "Priority",
     "Expected_Impact"]
] = performance.apply(
    generate_recommendation,
    axis=1
)

display(performance)
```

## Business Decision Report

```python
decision_report = performance[[
    "RegionSegment",
    "Total_Revenue",
    "Historical_CLTV",
    "Purchase_Frequency",
    "Recommendation",
    "Priority",
    "Expected_Impact"
]]

print("="*60)
print("BUSINESS DECISION REPORT")
print("="*60)

display(decision_report)
```

## High Priority Recommendations

```python
high_priority = decision_report[
    decision_report["Priority"] == "High"
]

print("="*60)
print("HIGH PRIORITY BUSINESS ACTIONS")
print("="*60)

display(high_priority)
```

##Business Summary

```python
summary = pd.DataFrame({

    "Category":[
        "Highest Revenue Region",
        "Highest Historical CLTV Region",
        "Highest Priority Recommendations"
    ],

    "Result":[

        performance.loc[
            performance["Revenue_Rank"]==1,
            "RegionSegment"
        ].values[0],

        performance.loc[
            performance["CLTV_Rank"]==1,
            "RegionSegment"
        ].values[0],

        len(high_priority)

    ]

})

display(summary)
```

##Save Report

```python
with pd.ExcelWriter("Business_Decision_Recommendations.xlsx") as writer:

    decision_report.to_excel(
        writer,
        sheet_name="Decision Report",
        index=False
    )

    high_priority.to_excel(
        writer,
        sheet_name="High Priority Actions",
        index=False
    )

    summary.to_excel(
        writer,
        sheet_name="Business Summary",
        index=False
    )

print("Business_Decision_Recommendations.xlsx saved successfully.")
```
