# Customer Value Segmentation Analysis

## Top 10 Customers by Revenue Analysis

**Objective:**
Identify the top 10 revenue-generating customers and visualize their revenue contribution using a bar chart.

```python
import pandas as pd

df = pd.read_excel("cleaned_online_retail.xlsx")

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

customer_revenue = (
    df.groupby("CustomerID")["Revenue"]
    .sum()
    .reset_index()
    .sort_values(by="Revenue", ascending=False)
)

print(customer_revenue.head(10))
```

##Graphical Representation of Top 10

```python
import matplotlib.pyplot as plt

top10 = customer_revenue.head(10)

plt.figure(figsize=(10,5))
plt.bar(top10["CustomerID"].astype(str), top10["Revenue"])

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Key Insights

- Customer 14646 generated the highest revenue.
- Revenue is concentrated among a small number of customers.
- High-value customers contribute significantly to overall business revenue.
- Retaining high-value customers can improve long-term profitability.

### Conclusion:
The analysis identified the top 10 revenue-generating customers and visualized their contribution through a bar chart. Understanding high-value customers helps businesses develop targeted retention strategies and maximize long-term revenue growth.
