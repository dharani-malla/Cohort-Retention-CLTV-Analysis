# DATA CLEANING

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("../data/online_retaildataset.xlsx")
```

```python
df.head()
```

# REMOVE MISSING CUSTOMER IDs

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['CustomerID'])
```

```python
df.isnull().sum()
```

```python
df.shape
```

# REMOVING CANCELLED TRANSACTIONS AND INVALID QUANTITIES 

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df.shape
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
df.head()
```

```python
df.shape
```

# CLEANED DATASET FOR COHORT ANALYSIS

```python
df.to_excel("cleaned_online_retail.xlsx", index=False)
```
