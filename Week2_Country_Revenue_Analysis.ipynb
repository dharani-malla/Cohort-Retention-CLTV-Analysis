# Country Revenue Analysis

**Objective:**
Analyze revenue generated across different countries and identify the top revenue-contributing countries using the cleaned e-commerce dataset.

```python
import pandas as pd

# Load dataset
df = pd.read_excel("cleaned_online_retail.xlsx")

# Create Revenue column
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

# Calculate country-wise revenue
country_revenue = (
    df.groupby("Country")["Revenue"]
    .sum()
    .reset_index()
    .sort_values(by="Revenue", ascending=False)
)

# Display Top 10 Countries
top10_countries = country_revenue.head(10)

print("TOP 10 COUNTRIES BY REVENUE")
print("-" * 50)
print(top10_countries)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(top10_countries["Country"], top10_countries["Revenue"])

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Key Insights

- The United Kingdom generated the highest revenue among all countries.
- A significant portion of total revenue is concentrated within a few countries.
- The top 10 countries contribute the majority of business revenue.
- Revenue share analysis helps identify the most profitable geographic markets.



## Conclusion

Country Revenue Analysis was performed to identify high-performing markets based on revenue generation. The analysis revealed the top revenue-contributing countries and their percentage share of total revenue. These insights can support business decisions related to market expansion, customer retention, and targeted marketing strategies.
