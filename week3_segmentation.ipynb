# CREATE REGION SEGMENTS

```python
sorted(df['Country'].unique())
```

```python
def region_segment(country):
    if country == "United Kingdom":
        return "UK"
    elif country in ["France", "Germany", "Spain", "Belgium", "Netherlands"]:
        return "Europe"
    else:
        return "Other"
df["RegionSegment"] = df["Country"].apply(region_segment)
```

```python
df[['Country', 'RegionSegment']].head(10)
```

```python
df['RegionSegment'].value_counts()
```
