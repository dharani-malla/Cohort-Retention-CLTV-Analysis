##Import Required Libraries

```python
import pandas as pd
```

##Load Business KPI Dashboard

```python
performance = pd.read_excel("Business_KPI_Dashboard.xlsx")
print("Dataset Loaded Successfully")
display(performance)
```

##Identify Best and Worst Performing Regions

```python
highest_revenue = performance.loc[performance["Total_Revenue"].idxmax()]
lowest_revenue = performance.loc[performance["Total_Revenue"].idxmin()]

highest_cltv = performance.loc[performance["Historical_CLTV"].idxmax()]
lowest_cltv = performance.loc[performance["Historical_CLTV"].idxmin()]

highest_aov = performance.loc[performance["Average_Order_Value"].idxmax()]
highest_pf = performance.loc[performance["Purchase_Frequency"].idxmax()]
```

##Executive Business Insights

```python
print("="*60)
print("EXECUTIVE BUSINESS INSIGHTS")
print("="*60)

print(f"\n1. Highest Revenue Region")
print(f"Region : {highest_revenue['RegionSegment']}")
print(f"Revenue: {highest_revenue['Total_Revenue']:,.2f}")
print("Insight: This region contributes the highest share of total revenue.")

print("\n" + "-"*60)

print(f"\n2. Highest Historical CLTV Region")
print(f"Region : {highest_cltv['RegionSegment']}")
print(f"Historical CLTV: {highest_cltv['Historical_CLTV']:.2f}")
print("Insight: Customers from this region generate the highest long-term value.")

print("\n" + "-"*60)

print(f"\n3. Lowest Revenue Region")
print(f"Region : {lowest_revenue['RegionSegment']}")
print(f"Revenue: {lowest_revenue['Total_Revenue']:,.2f}")
print("Insight: This region requires improved customer acquisition and retention.")

print("\n" + "-"*60)

print(f"\n4. Highest Average Order Value")
print(f"Region : {highest_aov['RegionSegment']}")
print(f"AOV : {highest_aov['Average_Order_Value']:.2f}")

print("\n" + "-"*60)

print(f"\n5. Highest Purchase Frequency")
print(f"Region : {highest_pf['RegionSegment']}")
print(f"Purchase Frequency : {highest_pf['Purchase_Frequency']:.2f}")
```

## Strategic Business Recommendations

```python
recommendations = []
recommendations.append(
    f"Strengthen customer loyalty initiatives in {highest_revenue['RegionSegment']} to maintain high revenue."
)
recommendations.append(
    f"Increase marketing investment in {highest_cltv['RegionSegment']} because it generates the highest customer lifetime value."
)
recommendations.append(
    f"Launch customer engagement campaigns in {lowest_revenue['RegionSegment']} to improve sales performance."
)
recommendations.append(
    "Use customer segmentation for targeted marketing and personalized offers."
)
recommendations.append(
    "Monitor customer retention and CLTV regularly to support business decisions."
)
print("="*60)
print("STRATEGIC BUSINESS RECOMMENDATIONS")
print("="*60)

for i, rec in enumerate(recommendations, 1):
    print(f"{i}. {rec}")
```

##  Executive Summary

```python
summary = pd.DataFrame({
    "Category": [
        "Highest Revenue Region",
        "Highest Historical CLTV",
        "Lowest Revenue Region",
        "Highest Average Order Value",
        "Highest Purchase Frequency"
    ],
    "Region": [
        highest_revenue["RegionSegment"],
        highest_cltv["RegionSegment"],
        lowest_revenue["RegionSegment"],
        highest_aov["RegionSegment"],
        highest_pf["RegionSegment"]
    ]
})

display(summary)
```

##Business Insights

```python
insights = pd.DataFrame({
    "Business Insight": [
        f"{highest_revenue['RegionSegment']} generated the highest revenue.",
        f"{highest_cltv['RegionSegment']} achieved the highest Historical CLTV.",
        f"{lowest_revenue['RegionSegment']} generated the lowest revenue.",
        f"{highest_aov['RegionSegment']} recorded the highest Average Order Value.",
        f"{highest_pf['RegionSegment']} recorded the highest Purchase Frequency."
    ],

    "Business Impact": [
        "Revenue generation is concentrated in this region.",
        "Customers from this region generate greater long-term value.",
        "Additional marketing efforts are required to improve revenue.",
        "Customers spend more per transaction.",
        "Customers purchase more frequently in this region."
    ]
})

display(insights)
```

## Save Executive Business Insights

```python
with pd.ExcelWriter("Executive_Business_Insights.xlsx") as writer:
    summary.to_excel(writer, sheet_name="Executive Summary", index=False)
    insights.to_excel(writer, sheet_name="Business Insights", index=False)
    recommendations_df = pd.DataFrame(recommendations, columns=["Recommendations"])
    recommendations_df.to_excel(writer, sheet_name="Recommendations", index=False)

print("Executive_Business_Insights.xlsx saved successfully.")
```
