#Historical CLTV Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CLTV results
cltv_df = pd.read_excel("Historical_CLTV_Performance_Analysis.xlsx")
```

##Segment-wise Historical CLTV

```python
# Display summary table
print("Segment-wise Historical CLTV")
print(cltv_df)

plt.figure(figsize=(8,5))
plt.bar(cltv_df["RegionSegment"], cltv_df["Historical_CLTV"])

plt.title("Segment-wise Historical CLTV Analysis")
plt.xlabel("Region")
plt.ylabel("Historical CLTV")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

##Highest & Lowest  CLTV Region

```python
highest = cltv_df.loc[cltv_df["Historical_CLTV"].idxmax()]
lowest = cltv_df.loc[cltv_df["Historical_CLTV"].idxmin()]

print("\nHighest CLTV Region")
print(highest)

print("\nLowest CLTV Region")
print(lowest)
```

##CLTV Comparison Table

```python
comparison = cltv_df.sort_values(by="Historical_CLTV", ascending=False)

print("\nCLTV Comparison")
print(comparison)
```
