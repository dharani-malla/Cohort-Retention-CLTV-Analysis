## Retention Trend Analysis

### Overall Retention Trends

### Observations

- Customer retention starts at 100% during the first purchase month.
- Retention drops significantly after the first month across almost all cohorts.
- Some cohorts maintain moderate retention (around 30%–45%) in later months.
- Most cohorts stabilize at lower retention levels over time.
- A few cohorts demonstrate better long-term customer retention than others.

### Key Findings

- Initial customer retention is very high during the first purchase month.
- A noticeable decline is observed after Month 1.
- A few cohorts maintain better retention than others.
- Customer engagement strategies can improve long-term retention.

## High Retention vs Low Retention Cohort Comparison

### Comparison

#### High Retention Cohorts

- Some customer cohorts maintain retention rates between 35% and 45% over multiple months.
- These cohorts indicate stronger customer loyalty and repeat purchases.
- Consistent engagement suggests effective customer retention.

#### Low Retention Cohorts

## Customer Churn Pattern Analysis

### Churn Analysis

#### Customer Churn Patterns

- Most customers stop making repeat purchases after the first month.

- Retention decreases gradually as the cohort age increases.
