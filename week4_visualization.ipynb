# RETENTION VISUALIZATION ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
retention_matrix = pd.read_excel("retention_matrix.xlsx",index_col=0)
```

# COHORT RETENTION HEATMAP

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    retention_matrix,
    annot=True,
    fmt=".0%",
    cmap="Blues"
)

plt.title("Customer Cohort Retention Heatmap")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Cohort Month")

plt.show()
```

# RETENTION DECAY LINE CHART

```python
plt.figure(figsize=(12,6))

for cohort in retention_matrix.index:
    plt.plot(
        retention_matrix.columns,
        retention_matrix.loc[cohort],
        marker='o'
    )

plt.title("Retention Decay by Cohort")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Retention Rate")
plt.legend(retention_matrix.index)

plt.show()
```
