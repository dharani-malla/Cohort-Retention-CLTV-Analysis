# Business Metrics Calculation

## Create Revenue Column (if not already present)

```python
if "Revenue" not in df.columns:
    df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("Revenue column is ready.")

df[["Quantity", "UnitPrice", "Revenue"]].head()
```

## Calculate Business Metrics by Region

```python
summary = df.groupby("RegionSegment").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("InvoiceNo", "nunique"),
    Total_Customers=("CustomerID", "nunique")
).reset_index()

summary
```

## Calculate Average Order Value (AOV)

```python
summary["Average_Order_Value"] = (
    summary["Total_Revenue"] /
    summary["Total_Orders"]
)

summary[["RegionSegment", "Average_Order_Value"]]
```

## Calculate Purchase Frequency

```python
summary["Purchase_Frequency"] = (
    summary["Total_Orders"] /
    summary["Total_Customers"]
)

summary["Average_Order_Value"] = summary["Average_Order_Value"].round(2)
summary["Purchase_Frequency"] = summary["Purchase_Frequency"].round(2)

summary
```
