# COHORT ANALYSIS PROJECT

```python
import pandas as pd
```

```python
df = pd.read_excel("cleaned_online_retail.xlsx")
```

```python
df.head()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['InvoiceDate'].dtype
```

# COHORT MONTH CREATION

```python
df['CohortMonth'] = df.groupby('CustomerID')['InvoiceDate'].transform('min').dt.to_period('M')
```

```python
df[['CustomerID','InvoiceDate','CohortMonth']].head()
```

```python
df['CohortMonth'].nunique()
```

# INVOICE MONTH CREATION

```python
df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')
```

```python
df[['CustomerID','InvoiceDate','CohortMonth','InvoiceMonth']].head()
```

```python
df['InvoiceMonth'].nunique()
```
