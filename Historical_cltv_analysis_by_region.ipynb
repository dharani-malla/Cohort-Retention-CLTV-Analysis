```python
import pandas as pd

# Load AOV & Purchase Frequency summary
df = pd.read_excel("AOV_PurchaseFrequency_Summary.xlsx")

# Calculate Historical CLTV
df["Historical_CLTV"] = (
    df["Average_Order_Value"] *
    df["Purchase_Frequency"]
).round(2)

print(df)

# Save to Excel
df.to_excel("Historical_CLTV_By_Region.xlsx", index=False)

print("Historical_CLTV_By_Region.xlsx saved successfully!")
```
