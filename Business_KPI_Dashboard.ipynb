##Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

##Load Dataset

```python
performance = pd.read_excel("Customer_Segment_Performance_Report.xlsx")
print("Dataset Loaded Successfully")
display(performance)
```

##Business KPI Summary

```python
total_revenue = performance["Total_Revenue"].sum()
total_customers = performance["Total_Customers"].sum()
total_orders = performance["Total_Orders"].sum()
average_aov = performance["Average_Order_Value"].mean()
average_pf = performance["Purchase_Frequency"].mean()
average_cltv = performance["Historical_CLTV"].mean()

print("="*50)
print("BUSINESS KPI DASHBOARD")
print("="*50)

print(f"Total Revenue               : {total_revenue:,.2f}")
print(f"Total Customers             : {total_customers}")
print(f"Total Orders                : {total_orders}")
print(f"Average Order Value         : {average_aov:.2f}")
print(f"Average Purchase Frequency  : {average_pf:.2f}")
print(f"Average Historical CLTV     : {average_cltv:.2f}")
```

##Revenue by Region

```python
plt.figure(figsize=(8,5))

plt.bar(performance["RegionSegment"],
        performance["Total_Revenue"])

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

##Historical CLTV by Region

```python
plt.figure(figsize=(8,5))

plt.bar(performance["RegionSegment"],
        performance["Historical_CLTV"])

plt.title("Historical CLTV by Region")
plt.xlabel("Region")
plt.ylabel("Historical CLTV")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

##Customer count by Region

```python
plt.figure(figsize=(8,5))

plt.bar(performance["RegionSegment"],
        performance["Total_Customers"])

plt.title("Customer Count by Region")
plt.xlabel("Region")
plt.ylabel("Customers")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

##Average Order Value by Region


```python
plt.figure(figsize=(8,5))

plt.bar(performance["RegionSegment"],
        performance["Average_Order_Value"])

plt.title("Average Order Value by Region")
plt.xlabel("Region")
plt.ylabel("Average Order Value")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

##Purchase Frequency by Region

```python
plt.figure(figsize=(8,5))

plt.bar(performance["RegionSegment"],
        performance["Purchase_Frequency"])

plt.title("Purchase Frequency by Region")
plt.xlabel("Region")
plt.ylabel("Purchase Frequency")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

##Excecutive Summary

```python
highest_revenue = performance.loc[performance["Total_Revenue"].idxmax()]
highest_cltv = performance.loc[performance["Historical_CLTV"].idxmax()]
highest_aov = performance.loc[performance["Average_Order_Value"].idxmax()]
highest_pf = performance.loc[performance["Purchase_Frequency"].idxmax()]

print("="*50)
print("EXECUTIVE SUMMARY")
print("="*50)

print(f"Highest Revenue Region           : {highest_revenue['RegionSegment']}")
print(f"Highest Historical CLTV Region   : {highest_cltv['RegionSegment']}")
print(f"Highest Average Order Value      : {highest_aov['RegionSegment']}")
print(f"Highest Purchase Frequency       : {highest_pf['RegionSegment']}")
```

```python
ranking = performance[[
    "RegionSegment",
    "Revenue_Rank",
    "CLTV_Rank"
]].sort_values("Revenue_Rank")

print("=" * 50)
print("REGION PERFORMANCE RANKING")
print("=" * 50)

display(ranking)
```

##Save the Dashboard

```python
performance.to_excel(
    "Business_KPI_Dashboard.xlsx",
    index=False
)

print("\nBusiness_KPI_Dashboard.xlsx saved successfully.")
```
