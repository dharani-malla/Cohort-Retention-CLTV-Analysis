# Cohort Performance Analysis



**Objective:**
Analyze cohort retention performance and identify the best and worst performing cohorts.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load retention matrix
retention = pd.read_excel("retention_matrix.xlsx")

print("Retention Matrix Shape:", retention.shape)

print("\nFirst 5 Rows")
print(retention.head())
```

## Calculate average retention for each cohort

```python
retention["Average_Retention"] = retention.iloc[:, 1:].mean(axis=1)
print("Average Retention by Cohort")
print(retention[["CohortMonth", "Average_Retention"]].head())
```

## Rank cohorts based on average retention

```python
cohort_ranking = retention.sort_values(
    by="Average_Retention",
    ascending=False
)
print("Cohort Performance Ranking")
print(cohort_ranking[["CohortMonth", "Average_Retention"]])
```

## Best and Worst Performing Cohorts


```python

best_cohort = cohort_ranking.iloc[0]
worst_cohort = cohort_ranking.iloc[-1]

print("Best Performing Cohort")
print(best_cohort)

print("\nWorst Performing Cohort")
print(worst_cohort)
```
