# Week 3 - Segment-wise Customer Behaviour Analysis

## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("default")
```

# Load Segmented Dataset

```python
df = pd.read_excel("Segmented_cohortdataset.xlsx")
df.head()
```

# Dataset Overview

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.isnull().sum()
```

# Business Metrics Calculation

## Create Revenue Column (if not already present)

```python
if "Revenue" not in df.columns:
    df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("Revenue column is ready.")

df[["Quantity", "UnitPrice", "Revenue"]].head()
```

## Calculate Business Metrics by Region

```python
summary = df.groupby("RegionSegment").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("InvoiceNo", "nunique"),
    Total_Customers=("CustomerID", "nunique")
).reset_index()

summary
```

## Calculate Average Order Value (AOV)

```python
summary["Average_Order_Value"] = (
    summary["Total_Revenue"] /
    summary["Total_Orders"]
)

summary[["RegionSegment", "Average_Order_Value"]]
```

## Calculate Purchase Frequency

```python
summary["Purchase_Frequency"] = (
    summary["Total_Orders"] /
    summary["Total_Customers"]
)

summary["Average_Order_Value"] = summary["Average_Order_Value"].round(2)
summary["Purchase_Frequency"] = summary["Purchase_Frequency"].round(2)

summary
```

## Purchasing Behaviour Comparison

```python
comparison = summary[
    [
        "RegionSegment",
        "Average_Order_Value",
        "Purchase_Frequency"
    ]
].sort_values(
    by="Average_Order_Value",
    ascending=False
)

comparison
```

### Interpretation

- Other region has the highest Average Order Value (AOV).
- Europe has a moderate Average Order Value.
- UK has the lowest Average Order Value among all regions.
- Purchase Frequency is almost similar across all regions.

# Export summary table to Excel

```python
summary.to_excel(
    "AOV_PurchaseFrequency_Summary.xlsx",
    index=False
)

print("Excel file exported successfully.")
```

# AOV & Purchase Frequency Visualizations

```python
plt.figure(figsize=(6,4))

plt.bar(
    summary["RegionSegment"],
    summary["Average_Order_Value"]
)

plt.title("Average Order Value by Region Segment")
plt.xlabel("Region Segment")
plt.ylabel("Average Order Value")

plt.show()
```

### Interpretation

- Other region has the highest Average Order Value (AOV).
- Europe has a moderate Average Order Value.
- UK has the lowest Average Order Value.

```python
plt.figure(figsize=(6,4))

plt.bar(
    summary["RegionSegment"],
    summary["Purchase_Frequency"]
)

plt.title("Purchase Frequency by Region Segment")
plt.xlabel("Region Segment")
plt.ylabel("Purchase Frequency")

plt.show()
```

### Interpretation

- Purchase Frequency is almost similar across all region segments.
- Other region has a slightly higher Purchase Frequency.
- Europe and UK have nearly equal Purchase Frequency.
