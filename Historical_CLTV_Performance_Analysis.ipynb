# CLTV Performance Analysis

```python
import pandas as pd
df = pd.read_excel("Historical_CLTV_By_Region.xlsx")
```

## 1. Rank Regions by Historical CLTV

```python
df = df.sort_values(by="Historical_CLTV", ascending=False).reset_index(drop=True)
df["Rank"] = df.index + 1
```

## 2. Calculate CLTV Contribution (%)

```python

total_cltv = df["Historical_CLTV"].sum()

df["CLTV_Contribution (%)"] = (
    df["Historical_CLTV"] / total_cltv * 100
).round(2)
```

## 3. Calculate Average CLTV

```python
average_cltv = round(df["Historical_CLTV"].mean(),2)
```

## 4. Difference from Average CLTV

```python
df["Difference_From_Average"] = (
    df["Historical_CLTV"] - average_cltv
).round(2)
```

## 5. Performance Status

```python
def performance(cltv):
    if cltv > average_cltv + 100:
        return "Above Average"
    elif cltv < average_cltv - 100:
        return "Below Average"
    else:
        return "Near Average"

df["Performance_Status"] = df["Historical_CLTV"].apply(performance)
```

## 6. Customer Value Classification

```python
highest = df["Historical_CLTV"].max()
lowest = df["Historical_CLTV"].min()

def customer_segment(cltv):
    if cltv == highest:
        return "High Value"
    elif cltv == lowest:
        return "Low Value"
    else:
        return "Medium Value"

df["Customer_Segment"] = df["Historical_CLTV"].apply(customer_segment)
```

## 7. Summary Statistics

```python
print("="*50)
print("Historical CLTV Summary")
print("="*50)

print(f"Total Historical CLTV : {total_cltv:.2f}")
print(f"Average Historical CLTV : {average_cltv:.2f}")
print(f"Highest Historical CLTV : {highest:.2f}")
print(f"Lowest Historical CLTV : {lowest:.2f}")

print("\n")
```

## 8. Final Summary Table

```python
summary = df[[
    "Rank",
    "RegionSegment",
    "Historical_CLTV",
    "CLTV_Contribution (%)",
    "Difference_From_Average",
    "Performance_Status",
    "Customer_Segment"
]]

print(summary)
```

## 9. Save to Excel

```python
summary.to_excel(
    "Historical_CLTV_Performance_Analysis.xlsx",
    index=False
)

print("\nExcel file saved successfully!")
```
