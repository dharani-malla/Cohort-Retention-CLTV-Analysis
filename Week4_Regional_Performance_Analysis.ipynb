```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Business KPI Dashboard

```python
performance = pd.read_excel("Business_KPI_Dashboard.xlsx")
display(performance)
```

## Retention and CLTV Trend Overview

```python
trend = performance[[
    "RegionSegment",
    "Total_Revenue",
    "Historical_CLTV",
    "Purchase_Frequency"
]]

print("="*60)
print("RETENTION AND CLTV TREND OVERVIEW")
print("="*60)

display(trend)
```

## Revenue vs Historical CLTV

```python
plt.figure(figsize=(8,5))

plt.scatter(
    performance["Total_Revenue"],
    performance["Historical_CLTV"]
)

for i in range(len(performance)):
    plt.text(
        performance["Total_Revenue"][i],
        performance["Historical_CLTV"][i],
        performance["RegionSegment"][i]
    )

plt.title("Revenue vs Historical CLTV")
plt.xlabel("Total Revenue")
plt.ylabel("Historical CLTV")

plt.grid(True)

plt.show()
```

##Purchase Frequency vs Historical CLTV

```python

plt.figure(figsize=(8,5))

plt.scatter(
    performance["Purchase_Frequency"],
    performance["Historical_CLTV"]
)

for i in range(len(performance)):
    plt.text(
        performance["Purchase_Frequency"][i],
        performance["Historical_CLTV"][i],
        performance["RegionSegment"][i]
    )

plt.title("Purchase Frequency vs Historical CLTV")
plt.xlabel("Purchase Frequency")
plt.ylabel("Historical CLTV")

plt.grid(True)

plt.show()
```

##Trend Analysis

```python


highest_revenue = performance.loc[
    performance["Total_Revenue"].idxmax()
]

highest_cltv = performance.loc[
    performance["Historical_CLTV"].idxmax()
]

highest_pf = performance.loc[
    performance["Purchase_Frequency"].idxmax()
]

print("="*60)
print("TREND ANALYSIS")
print("="*60)

print(f"""
Revenue Trend
-------------
{highest_revenue['RegionSegment']} generated the highest revenue,
indicating that this region is the primary revenue contributor.

Historical CLTV Trend
---------------------
{highest_cltv['RegionSegment']} achieved the highest Historical CLTV,
showing stronger long-term customer value.

Purchase Frequency Trend
------------------------
{highest_pf['RegionSegment']} recorded the highest purchase frequency,
indicating better customer engagement.
""")
```

##Business Observations

```python
observations = pd.DataFrame({
    "Observation": [
        "Highest Revenue Region",
        "Highest Historical CLTV Region",
        "Highest Purchase Frequency Region",
        "Business Conclusion"
    ],
    "Result": [
        f"{highest_revenue['RegionSegment']} generated the highest total revenue, making it the primary revenue contributor.",
        f"{highest_cltv['RegionSegment']} achieved the highest Historical CLTV, indicating greater long-term customer value.",
        f"{highest_pf['RegionSegment']} recorded the highest purchase frequency, reflecting stronger customer engagement and repeat purchases.",
        f"Regional performance analysis indicates that {highest_revenue['RegionSegment']} drives the highest revenue, while {highest_cltv['RegionSegment']} delivers the highest customer lifetime value. Businesses should focus on retaining customers in {highest_revenue['RegionSegment']} and expanding customer acquisition efforts in {highest_cltv['RegionSegment']} to maximize long-term business growth."
    ]

})
print("=" * 60)
print("BUSINESS OBSERVATIONS")
print("=" * 60)
display(observations)
```

##Regional Comparison

```python
comparison = pd.DataFrame({

    "Region": ["UK", "Other", "Europe"],

    "Strength": [
        "Highest Revenue",
        "Highest Historical CLTV & Purchase Frequency",
        "Balanced Customer Performance"
    ],

    "Weakness": [
        "Lower Historical CLTV",
        "Smaller Customer Base",
        "Lowest Revenue"
    ],

    "Suggested Action": [
        "Improve customer retention and loyalty programs",
        "Increase customer acquisition campaigns",
        "Launch promotional campaigns to improve revenue"
    ]

})

print("="*60)
print("REGIONAL COMPARISON")
print("="*60)

display(comparison)
```

Save Regional Performance Report

```python
with pd.ExcelWriter("Regional_Performance_Analysis.xlsx") as writer:

    trend.to_excel(
        writer,
        sheet_name="Regional Metrics",
        index=False
    )

    observations.to_excel(
        writer,
        sheet_name="Business Observations",
        index=False
    )

    comparison.to_excel(
        writer,
        sheet_name="Regional Comparison",
        index=False
    )

print("Regional_Performance_Analysis.xlsx saved successfully.")
```

## Conclusion

The regional performance analysis compared customer segments using revenue, Historical CLTV, and purchase frequency.

Key findings:
- UK is the highest revenue-generating region.
- Other has the highest Historical CLTV and purchase frequency.
- Europe has growth potential through targeted marketing initiatives.

These findings support strategic decisions for customer retention, acquisition, and long-term business growth.
