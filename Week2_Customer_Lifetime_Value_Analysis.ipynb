# Customer Lifetime Value (CLTV) Analysis

**Objective:**
Calculate Customer Lifetime Value (CLTV) using Average Order Value (AOV) and Purchase Frequency to estimate the long-term value generated by customers.

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_excel("cleaned_online_retail.xlsx")

# Create Revenue column
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

# Calculate AOV
total_revenue = df["Revenue"].sum()
total_orders = df["InvoiceNo"].nunique()
aov = total_revenue / total_orders

# Calculate Purchase Frequency
total_customers = df["CustomerID"].nunique()
purchase_frequency = total_orders / total_customers

# Calculate CLTV
cltv = aov * purchase_frequency

results = pd.DataFrame({
    "Metric": ["AOV", "Purchase Frequency", "CLTV"],
    "Value": [round(aov, 2), round(purchase_frequency, 2), round(cltv, 2)]
})

print("\nCLTV ANALYSIS RESULTS")
print("-" * 40)
print(results.to_string(index=False))
```

**Conclusion:**
Customer Lifetime Value (CLTV) was estimated using Average Order Value (AOV) and Purchase Frequency. The resulting CLTV metric provides an estimate of the value generated by customers and helps identify the long-term revenue potential of the customer base.
